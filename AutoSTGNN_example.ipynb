## Training examples

### Convergence

The following code exposes adjustable parameters so that the user can change architecture-dataset combinations as well as number of configs ran for how many epochs

````python
from convergence import run_convergence_analysis

run_convergence_analysis(
    model_name='graphwavenet',
    dataset_name='metrla',
    n_configs=2,
    max_epochs=1,
    base_root='/content/drive/MyDrive/AutoTest/data',
    output_root='/content/drive/MyDrive/AutoTest/convergence_plots',
    seed = 42,
)
````

### Graph Wavenet on Metr-LA
The following cell shows how to run random search and BO-DE for Graph Wavenet-Metr-LA. The budgets for each are also adjustable. Both RS and BO-DE are able to resume crashed runs through passing the json to the functions as shown below. 

````python
from random_search import run_random_search

run_random_search(
    model_name='graphwavenet',
    dataset_name='metrla',
    n_trials=1,
    max_epochs=1,
    results_dir='/content/drive/MyDrive/AutoTest/search_results',
    base_root='/content/drive/MyDrive/AutoTest/data',
#   resume_from= "/content/drive/MyDrive/AutoTest/search_results/x.json" filepath to json for resuming crashed runs


)
````

````python
from bo_de import run_bode

log = run_bode(
    model_name='graphwavenet',
    dataset_name='metrla',
    T=1, # total bode runs
    n_init=1, # total initiliasation runs
    # DE Params
    n_pop=60,
    k=100,
    f=0.5,
    p_c=0.9,
    
    max_epochs=1,
    train_seed=42,
    results_dir='/content/drive/MyDrive/AutoTest/search_results',
    base_root='/content/drive/MyDrive/AutoTest/data',
#   resume_from= "/content/drive/MyDrive/AutoTest/search_results/x.json" filepath to json for resuming crashed runs
)
````
